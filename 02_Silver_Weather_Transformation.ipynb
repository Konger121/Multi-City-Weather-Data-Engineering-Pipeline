```python
from pyspark.sql import functions as F

silver_df = (
    spark.table("weather_bronze")
    .withColumn("timestamp", F.to_timestamp("timestamp"))
    .withColumn("date", F.to_date("timestamp"))
    .withColumn("hour", F.hour("timestamp"))
    .withColumn("day_of_week", F.dayofweek("timestamp"))
    .withColumn("month", F.month("timestamp"))
)
```

```python
silver_df.select(
    F.count("*").alias("total_records"),
    F.sum(F.when(F.col("city").isNull(), 1).otherwise(0)).alias("null_city"),
    F.sum(F.when(F.col("timestamp").isNull(), 1).otherwise(0)).alias("null_timestamp"),
    F.sum(F.when(F.col("temperature_2m").isNull(), 1).otherwise(0)).alias("null_temperature"),
    F.sum(F.when(F.col("relative_humidity_2m").isNull(), 1).otherwise(0)).alias("null_humidity"),
    F.sum(F.when(F.col("precipitation").isNull(), 1).otherwise(0)).alias("null_precipitation"),
    F.sum(F.when(F.col("wind_speed_10m").isNull(), 1).otherwise(0)).alias("null_wind_speed")
).display()
```

```python
silver_df.write.format("delta").mode("overwrite").saveAsTable("weather_silver")
```
