```python
spark.sql("""
SELECT
    COUNT(*) AS total_records,
    COUNT(DISTINCT f.location_id) AS cities,
    MIN(f.timestamp) AS earliest_timestamp,
    MAX(f.timestamp) AS latest_timestamp
FROM fact_weather f
JOIN dim_location l
    ON f.location_id = l.location_id
""").display()
```

```python
spark.sql("""
SELECT
    l.city,
    COUNT(*) AS readings,
    ROUND(AVG(f.temperature), 2) AS avg_temperature,
    ROUND(MIN(f.temperature), 2) AS min_temperature,
    ROUND(MAX(f.temperature), 2) AS max_temperature,
    ROUND(AVG(f.humidity), 2) AS avg_humidity,
    ROUND(SUM(f.precipitation), 2) AS total_precipitation,
    ROUND(AVG(f.wind_speed), 2) AS avg_wind_speed
FROM fact_weather f
JOIN dim_location l
    ON f.location_id = l.location_id
GROUP BY l.city
ORDER BY avg_temperature DESC
""").display()
```

```python
spark.sql("""
SELECT
    l.city,
    d.date,
    ROUND(AVG(f.temperature), 2) AS avg_temperature,
    ROUND(MIN(f.temperature), 2) AS min_temperature,
    ROUND(MAX(f.temperature), 2) AS max_temperature,
    ROUND(AVG(f.humidity), 2) AS avg_humidity,
    ROUND(SUM(f.precipitation), 2) AS total_precipitation,
    ROUND(AVG(f.wind_speed), 2) AS avg_wind_speed
FROM fact_weather f
JOIN dim_location l
    ON f.location_id = l.location_id
JOIN dim_date d
    ON f.date_id = d.date_id
GROUP BY l.city, d.date
ORDER BY d.date, l.city
""").display()
```

```python
spark.sql("""
SELECT
    l.city,
    f.timestamp,
    f.temperature,
    f.humidity,
    f.precipitation,
    f.wind_speed
FROM fact_weather f
JOIN dim_location l
    ON f.location_id = l.location_id
ORDER BY f.temperature DESC
LIMIT 10
""").display()
```

```python
spark.sql("""
SELECT
    l.city,
    ROUND(AVG(f.temperature), 2) AS avg_temperature,
    ROUND(AVG(f.humidity), 2) AS avg_humidity,
    ROUND(AVG(f.wind_speed), 2) AS avg_wind_speed,
    ROUND(SUM(f.precipitation), 2) AS total_precipitation
FROM fact_weather f
JOIN dim_location l
    ON f.location_id = l.location_id
GROUP BY l.city
ORDER BY total_precipitation DESC
""").display()
```

```python
spark.sql("""
WITH city_summary AS (
    SELECT
        l.city,
        AVG(f.temperature) AS avg_temperature
    FROM fact_weather f
    JOIN dim_location l
        ON f.location_id = l.location_id
    GROUP BY l.city
)

SELECT
    city,
    ROUND(avg_temperature, 2) AS avg_temperature,
    RANK() OVER (
        ORDER BY avg_temperature DESC
    ) AS temperature_rank
FROM city_summary
ORDER BY temperature_rank
""").display()
```
